```python
df = pd.read_csv("/kaggle/input/datasets/janiobachmann/bank-marketing-dataset/bank.csv")
df
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
df.apply(lambda x:print(np.unique(x,return_counts=True)), axis=0)
```

```python
x = df[df.columns.drop(["deposit"])]
y = df["deposit"]
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(x,y,test_size=0.2, random_state=42)
# xtrain, xtest, ytrain, ytest
```

```python
ytrain
```

```python
#p(y)
py = len(ytrain[ytrain=="yes"])/len(ytrain)
py
```

```python
num = ["age", "balance","day","duration","campaign","pdays", "previous"]
```

$
p(x|y) = \frac{1}{\sqrt{2 \pi} \sigma} e^{- \frac{(x-\mu)^2}{2 \sigma^2}}
$

```python
pxy = pd.concat({col:xtrain[col].loc[ytrain=="yes"].value_counts()/len(ytrain[ytrain=="yes"]) if col not in num else pd.Series({"var":np.var(xtrain[col].loc[ytrain=="yes"]), "mu":np.mean(xtrain[col].loc[ytrain=="yes"])})  for col in xtrain.columns}).rename("count")
pxy
```

```python
pxy.groupby(level=0).sum()
```

```python
#p(y|x)
pyx = xtest.apply(lambda row: np.log(py) + sum(np.log(pxy[col][row[col]] if col not in num else ((1/np.sqrt(2*np.pi*pxy[col]["var"])) * np.exp(-((row[col]-pxy[col]["mu"])**2)/(2*pxy[col]["var"])))) for col in row.index), axis=1)
pyx
```

```python
#p(n)
pn = len(ytrain[ytrain=="no"])/len(ytrain)
pn
```

```python
pxn = pd.concat({col:xtrain[col].loc[ytrain=="no"].value_counts()/len(ytrain[ytrain=="no"]) if col not in num else pd.Series({"var":np.var(xtrain[col].loc[ytrain=="no"]), "mu":np.mean(xtrain[col].loc[ytrain=="no"])})  for col in xtrain.columns}).rename("count")
pxn
```

```python
#p(n|x)
pnx = xtest.apply(lambda row: np.log(pn) + sum(np.log(pxn[col][row[col]] if col not in num else ((1/np.sqrt(2*np.pi*pxn[col]["var"])) * np.exp(-((row[col]-pxn[col]["mu"])**2)/(2*pxn[col]["var"])))) for col in row.index), axis=1)
pnx
```

```python
pred = pyx>pnx
pred
```

```python
yhat = pd.Series(np.where(pred, "yes", "no"))
yhat
```

```python
accuracy_score(yhat, ytest)
```
